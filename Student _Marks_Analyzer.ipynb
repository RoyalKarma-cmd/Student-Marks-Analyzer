```python
marks = {
    "Maths" : 75,
    "Physics" : 68,
    "Science" : 80,
    "English" : 30,
}                       # Storing the values 

for subjects, mark in marks.items():

    if mark >= 90:
        print(subjects, ":", mark, "-", "A+")

    elif mark >= 80:
        print(subjects, ":", mark, "-", "B+")

    elif mark >= 65:
        print(subjects, ":", mark, "-", "C+")

    elif mark >= 50:
        print(subjects, ":", mark, "-", "D+")
    
    else:
        print(subjects, ":", mark, "-", "FAIL!")

```
